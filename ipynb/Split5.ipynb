# Testing

````python
import pickle

maximum_top1_acc = test_accs_top1.index(max(test_accs_top1))

class_report, conf_matrix = loop(root="frames", 
                                index_cache="frames_index.txt", 
                                splits_file="splits.pkl", 
                                get_splits=f"split{split}_test", 
                                Is_train=False, 
                                local_model_path=f'{result_path}/Split{split}_{maximum_top1_acc}.pth',
                                epochs=False,
                                model_name=f'Split{split}', 
                                mapping='./mapping.txt')


results_file_pkl = f'{result_path}/split{split}_results.pkl'

results_dict = {
  'classification_report': class_report,
  'confusion_matrix': conf_matrix}

with open(results_file_pkl, 'wb') as f:
  pickle.dump(results_dict, f)
````

---

# Extracting per-class ACC

````python
import pandas as pd

with open('./mapping.txt', 'r') as f:
    class_mappings = f.readlines()
class_names = [line.strip().split(' ', 1)[1] for line in class_mappings]
total_samples = conf_matrix.sum()

per_class_accuracy = {}
for idx, class_name in enumerate(class_names):
    TP = conf_matrix[idx, idx]  
    total_samples_per_class = conf_matrix[idx, :].sum()  # All samples of this class

    # Avoid division by zero
    if total_samples_per_class > 0:
        acc = TP / total_samples_per_class
    else:
        acc = 0  # If there are no samples, set accuracy to 0

    per_class_accuracy[class_name] = acc

    if class_name in class_report:
        class_report[class_name]['Accuracy'] = acc
  

class_report_pd = pd.DataFrame(class_report).T 
class_report_pd.columns=['precision', 'recall', 'f1-score', 'support', 'Accuracy']

class_report_pd.to_csv(f'{result_path}/Split{split}_report.csv')
````
